```python
from ultralytics import YOLO

model = YOLO("best.pt")
```

```python
filename = list(uploaded.keys())[0]

results = model.predict(
    source=filename,
    save=True,
    conf=0.4
)
```

```python
from PIL import Image
import os

output = os.path.join(results[0].save_dir, os.path.basename(results[0].path))
Image.open(output)
```
